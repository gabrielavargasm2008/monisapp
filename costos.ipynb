# Costo de una ruta de entrega

Este notebook calcula cuánto cuesta una ruta de entrega.
Cambia un valor de la celda de parámetros y mira cómo cambia el costo.

```python
# Parámetros de la ruta
valor_km = 1800
peaje = 3444
km = 10
```

```python
costo = km * valor_km + peaje
print(f"Costo de la ruta de {km} km: ${costo:,}")
```
